Task3:


This is a regression problem because the target, SalePrice, is a continuous numeric value rather than a discrete category or class label , the model needs to predict "how much," not "which class." This changes evaluation later: instead of accuracy/precision/recall (used for classification), we need metrics that measure numeric distance between predicted and actual values, like MAE, MSE, RMSE, and R²

```python
# task4:

df_subset['CentralAir'] = df_subset['CentralAir'].map({'Y': 1, 'N': 0})

df_encoded = p.get_dummies(df_subset, 
                             columns=['Neighborhood', 'HouseStyle', 'KitchenQual'], 
                             drop_first=True)

df_encoded.head()
```

interpretation:


CentralAir was mapped directly to 0/1 since it's a strictly binary Yes/No field  no information is lost by using a single column. Neighborhood, HouseStyle, and KitchenQual were one-hot encoded instead of label-encoded (e.g. 1,2,3...) because they have no inherent order assigning numbers would falsely imply that, say, one neighborhood is "greater than" another, which would mislead the linear model into treating category identity as a magnitude.

TASK 10 :

The intercept does not have a sensible real-world interpretation for this dataset. It represents the predicted house price when all input features are zero, but a house with all features equal to zero is not a realistic situation. Therefore, the intercept is mainly useful as part of the mathematical regression equation rather than as a meaningful standalone house-price estimate.

```python
# task 11 :

import matplotlib.pyplot as plt

residuals = y_test - pre


plt.figure(figsize=(8,5))
plt.scatter(pre, residuals, alpha=0.5)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Predicted Price")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residuals vs Predicted Values")
plt.show()

plt.figure(figsize=(8,5))
plt.hist(residuals, bins=30, edgecolor='black')
plt.xlabel("Residual")
plt.ylabel("Frequency")
plt.title("Distribution of Residuals")
plt.show()

print("Mean residual:", residuals.mean())
```

interpretation:
The residuals are not randomly scattered , there is a clear visible pattern.

The residuals vs predicted plot shows a funnel (cone) shape: for predicted prices below about $150,000 the residuals are tightly clustered around zero, but as predicted price increases beyond $200,000 they spread out sharply, reaching errors of over ±$100,000. This is heteroscedasticity ,the variance of the errors increases with price , which violates the constant-variance assumption of Linear Regression. Practically, it means the model predicts cheap and mid-range houses reasonably well but becomes unreliable for expensive ones. There is also a slight tendency to overpredict in the $250,000–$350,000 range, where more points fall below the zero line than above it.

The histogram is roughly bell-shaped and centered near zero, which is expected, but it is right-skewed with a long tail of large positive residuals extending past +$150,000 and +$250,000, with no matching negative tail. These are houses that sold for far more than the model predicted, meaning the model fails to capture whatever drives prices at the very top of the market , likely features I excluded from my 10–12 feature subset, such as lot area, exterior quality, or renovation status.

Overall, the residual pattern suggests the relationship between my features and SalePrice is not purely linear, and that a log transformation of the target would likely stabilise the variance.
